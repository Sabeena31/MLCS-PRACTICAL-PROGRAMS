```python
import pandas as pd
from sklearn.svm import SVC

data = pd.DataFrame({
    "url_len": [18, 22, 45, 50, 20, 25, 48, 55],
    "dots": [1, 2, 5, 4, 1, 2, 5, 6],
    "https": [1, 1, 0, 0, 1, 1, 0, 0],
    "label": [0, 0, 1, 1, 0, 0, 1, 1]
})

X = data[["url_len", "dots", "https"]]
y = data["label"]

model = SVC(kernel="linear")
model.fit(X, y)

pred = model.predict(X)

print("Predicted labels:", pred)
print("Accuracy:", model.score(X, y))
```
