```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

data = {
    "packets": [120, 150, 130, 110, 5000, 7000],
    "bytes": [5000, 6000, 5500, 5200, 80000, 120000],
    "duration": [10, 12, 11, 9, 2, 1],
    "connections": [5, 6, 5, 4, 500, 800]
}

df = pd.DataFrame(data)

scaler = StandardScaler()
X = scaler.fit_transform(df)

model = KMeans(n_clusters=2, random_state=42, n_init=10)
df["cluster"] = model.fit_predict(X)

centroids = scaler.inverse_transform(model.cluster_centers_)
normal = centroids[:, 0].argmin()

df["category"] = df["cluster"].apply(
    lambda x: "Normal" if x == normal
    else "Abnormal"
)

print(df)
```
