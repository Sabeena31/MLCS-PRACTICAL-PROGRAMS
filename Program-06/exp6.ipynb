```python
import pandas as pd
from urllib.parse import urlparse
from sklearn.ensemble import IsolationForest

urls = [
    "https://google.com",
    "https://example.com",
    "https://github.com",
    "http://safe-site.com",
    "http://abc.com/login",
    "http://very-long-suspicious-site.com/a/b/c/d/e/f"
]

def features(url):
    return [
        len(url),
        url.count("."),
        url.count("/"),
        sum(c in "@?=&%" for c in url)
    ]

X = pd.DataFrame(
    [features(u) for u in urls],
    columns=["length", "dots", "slashes", "special"]
)

model = IsolationForest(contamination=0.2, random_state=42)
X["anomaly"] = model.fit_predict(X)

for url, result in zip(urls, X["anomaly"]):
    print(url, "->", "Anomaly" if result == -1 else "Normal")
```
